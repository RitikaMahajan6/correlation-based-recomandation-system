# Data Exploration for Correlation-Based Recommendation System

This notebook is used for exploratory data analysis (EDA) on the dataset used in the correlation-based recommendation system. The goal is to understand the data better, visualize key features, and derive insights that will inform the model training process.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
data_path = '../data/raw/dataset.csv'
df = pd.read_csv(data_path)

# Display the first few rows of the dataset
df.head()
```

```python
# Summary statistics
df.describe()
```

```python
# Check for missing values
missing_values = df.isnull().sum()
missing_values[missing_values > 0]
```

```python
# Visualize the distribution of key features
plt.figure(figsize=(12, 6))
sns.histplot(df['feature_name'], bins=30, kde=True)
plt.title('Distribution of Feature Name')
plt.xlabel('Feature Name')
plt.ylabel('Frequency')
plt.show()
```

```python
# Correlation matrix
plt.figure(figsize=(10, 8))
correlation_matrix = df.corr()
sns.heatmap(correlation_matrix, annot=True, fmt='.2f', cmap='coolwarm', square=True)
plt.title('Correlation Matrix')
plt.show()
```

## Insights
Based on the exploratory data analysis, we can derive insights that will help in the model training process. Key features and their relationships will guide the selection of algorithms and parameters.
